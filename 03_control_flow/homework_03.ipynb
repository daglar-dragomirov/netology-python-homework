## Задание 1

Дана переменная `word`, в которой хранится слово. Нужно вывести:
- среднюю букву, если число букв в слове нечётное;
- две средние буквы, если число букв чётное.

```python
def middle_letters(word):
    """Возвращает среднюю букву (нечётная длина) или две средние буквы (чётная длина)."""
    middle = len(word) // 2
    if len(word) % 2 == 0:
        return word[middle - 1:middle + 1]
    return word[middle]
```

```python
word = 'test'
print(middle_letters(word))
```

```python
word = 'testing'
print(middle_letters(word))
```

## Задание 2 (необязательное)

MVP dating-сервиса: сортируем имена юношей и девушек по алфавиту и знакомим людей с одинаковыми индексами. Если списки разной длины, кто-то может остаться без пары, и тогда никого не знакомим.

```python
def print_perfect_pairs(boys, girls):
    if len(boys) != len(girls):
        print('Внимание, кто-то может остаться без пары!')
        return
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(f'{boy} и {girl}')
```

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']
print_perfect_pairs(boys, girls)
```

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']
print_perfect_pairs(boys, girls)
```
