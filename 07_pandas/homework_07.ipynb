## Задание 1

Определяем, какому фильму выставили больше всего оценок 5.0.

```python
movies = pd.read_csv('movies.csv')
ratings = pd.read_csv('ratings.csv')

five_star_counts = (
    ratings[ratings['rating'] == 5.0]
    .groupby('movieId')
    .size()
    .rename('five_star_ratings')
    .reset_index()
    .merge(movies[['movieId', 'title']], on='movieId')
    .sort_values('five_star_ratings', ascending=False)
)
five_star_counts.head()
```

```python
top_movie = five_star_counts.iloc[0]
print(f"Больше всего оценок 5.0 у фильма «{top_movie['title']}»: {top_movie['five_star_ratings']}")
```

## Задание 2

Суммарное потребление стран Прибалтики (Латвия, Литва, Эстония) категорий 4, 12 и 21 за 2005–2010 годы без учёта отрицательных значений `quantity`.

```python
power = pd.read_csv('power.csv')

baltic = power[
    power['country'].isin(['Latvia', 'Lithuania', 'Estonia'])
    & power['category'].isin([4, 12, 21])
    & power['year'].between(2005, 2010)
    & (power['quantity'] >= 0)
]

total = baltic['quantity'].sum()
print(f'Суммарное потребление: {total}')
```

Проверка: разбивка по странам и категориям.

```python
baltic.pivot_table(index='country', columns='category', values='quantity', aggfunc='sum', margins=True, margins_name='Итого')
```
