Программа для секретаря: по командам пользователя ищет данные о документах.

Команды:
- `p` — узнать владельца документа по его номеру;
- `s` — узнать, на какой полке хранится документ;
- `q` — завершить работу.

Глобальные переменные — только `documents` и `directories`, ввод обрабатывается в цикле `while` до команды `q`.

```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}
```

```python
def ask_document_number():
    return input('Введите номер документа:\n').strip()


def find_owner(doc_number):
    """Возвращает имя владельца документа или None, если документа нет."""
    for document in documents:
        if document['number'] == doc_number:
            return document['name']
    return None


def find_shelf(doc_number):
    """Возвращает номер полки с документом или None, если документ не найден."""
    for shelf, numbers in directories.items():
        if doc_number in numbers:
            return shelf
    return None


def show_owner():
    owner = find_owner(ask_document_number())
    print('Результат:')
    print(f'Владелец документа: {owner or "владелец не найден"}')


def show_shelf():
    shelf = find_shelf(ask_document_number())
    print('Результат:')
    if shelf is None:
        print('Документ не найден.')
    else:
        print(f'Документ хранится на полке: {shelf}')


def main():
    commands = {'p': show_owner, 's': show_shelf}
    while True:
        command = input('Введите команду:\n').strip()
        if command == 'q':
            print('Работа завершена.')
            break
        action = commands.get(command)
        if action is None:
            print('Неизвестная команда. Доступные команды: p, s, q')
            continue
        action()
```

Запуск программы. Проверяем все случаи из чек-листа: существующий и несуществующий номер для каждой команды, затем выход по `q`.

```python
main()
```
